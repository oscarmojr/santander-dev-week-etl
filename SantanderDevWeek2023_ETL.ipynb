Desafio DIO: Extração, Transformação e Carregamento com Python.
Como a API original pode estar fora do ar, os dados vêm de um CSV (Alternativa 2).

## Extração
Lemos o arquivo `data/usuarios.csv` com as colunas Nome, Conta e Cartão.

```python
from pathlib import Path
import etl

df = etl.extrair(Path('data/usuarios.csv'))
df
```

## Transformação
Geramos uma mensagem personalizada sobre investimentos para cada usuário.
Se `OPENAI_API_KEY` estiver configurada, a mensagem vem da IA; senão, de um gerador local.

```python
df_transformado = etl.transformar(df)
df_transformado[['Nome', 'Mensagem']]
```

## Carregamento
Salvamos o resultado em CSV e JSON na pasta `output/`.

```python
etl.carregar(df_transformado, Path('output'))
print(Path('output/usuarios_com_mensagens.json').read_text(encoding='utf-8')[:400])
```
